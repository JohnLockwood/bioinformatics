# Reading the human Genome GTF File

This file is in Gene Transfer format. Per [wikipedia](https://en.wikipedia.org/wiki/Gene_transfer_format):

> The **Gene transfer format** (**GTF**) is a [file format](https://en.wikipedia.org/wiki/File_format "File format") used to hold information about [gene](https://en.wikipedia.org/wiki/Gene "Gene") structure. It is a [tab-delimited](https://en.wikipedia.org/wiki/Tab_key "Tab key") text format based on the [general feature format](https://en.wikipedia.org/wiki/General_feature_format "General feature format") (GFF), but contains some additional conventions specific to gene information. A significant feature of the GTF that can be validated: given a sequence and a GTF file, one can check that the format is correct. This significantly reduces problems with the interchange of data between groups.:

````python
from gtfparse import read_gtf
import os
import polars as pl

def getfile():
    dbdir = os.environ["BIO_DATA"]
    filename = os.path.join(dbdir, "hg38.ncbiRefSeq.gtf")
    return filename

df = read_gtf(getfile())
````

````python
dbdir = os.environ["BIO_DATA"]
parquet_file = os.path.join(dbdir, "hg38.ncbiRefSeq.parquet")

df.write_parquet(parquet_file)
````
